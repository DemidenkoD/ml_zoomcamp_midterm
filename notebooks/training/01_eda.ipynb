### 💡 Useful Functions & API Hints:
- **Correlation Matrix:** Call `df[columns_list].corr()` to calculate pairwise Pearson correlation coefficients (values between -1 and +1).
- **Heatmap Visualization (`sns.heatmap`):**
  - `annot=True`: Displays the numerical correlation values inside each cell.
  - `cmap='coolwarm'`: A diverging colormap showing negative correlations in blue and positive in red.
  - `fmt='.2f'`: Formats numbers to 2 decimal places.
  - `vmin=-1, vmax=1`: Sets the color scale bounds from -1 to +1.
- **Extracting a single correlation value:** You can slice a correlation DataFrame using `.loc['survey_stress_score', 'wearable_stress_score']` or `.iloc[0, 1]` on a 2x2 correlation matrix.
- **Categorical Scatter Plot (`sns.stripplot`):**
  - `jitter=0.25`: Adds horizontal spread to points so overlapping dots are visible.
  - `alpha=0.5`: Adds transparency to prevent occlusion.

### 🛡️ Why Exclude `wearable_stress_score` and `user_id`?

If included, the model will just bypass lifestyle factors and read the wearable metric. Furthermore, our final Streamlit web application is designed for users answering a lifestyle questionnaire without requiring wearable sensors.
- **Identifier Leakage:** `user_id` is an arbitrary participant index with no generalizable relationship to stress.
